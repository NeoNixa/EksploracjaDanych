## Analiza zmiennej docelowej (SalePrice)

```python
# min, max, srednia, mediana, Q1, Q3, IQR
sp = df['SalePrice']
q1 = sp.quantile(0.25)
q3 = sp.quantile(0.75)
iqr = q3 - q1

print("Min:", sp.min())
print("Max:", sp.max())
print("Srednia:", sp.mean())
print("Mediana:", sp.median())
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

```python
# histogram cen dla roznych bins (10, 30, 50)
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.hist(df['SalePrice'], bins=10, edgecolor='black')
plt.title('bins = 10')
plt.xlabel('SalePrice')

plt.subplot(1, 3, 2)
plt.hist(df['SalePrice'], bins=30, edgecolor='black')
plt.title('bins = 30')
plt.xlabel('SalePrice')

plt.subplot(1, 3, 3)
plt.hist(df['SalePrice'], bins=50, edgecolor='black')
plt.title('bins = 50')
plt.xlabel('SalePrice')

plt.tight_layout()
plt.show()
```

Rozkład cen nie jest normalny ani symetryczny – jest wyraźnie skośny w prawo (długi ogon po prawej stronie). Średnia (~180k) jest zauważalnie wyższa od mediany (163k), bo kilka bardzo drogich domów podbija średnią.

```python
# boxplot cen
plt.figure(figsize=(8, 3))
sns.boxplot(x=df['SalePrice'])
plt.title('Wykres pudełkowy cen')
plt.show()
```

Na wykresie pudełkowym widać sporo wartości odstających (powyżej ~340k $). Najdroższy dom kosztuje 755k $. To po prostu luksusowe wille o bardzo wysokim standardzie wykończenia.

## Co wpływa na cenę domu?

```python
# kolumny zwiazane z powierzchnia (Area i SF)
area_cols = [c for c in df.columns if 'Area' in c or 'SF' in c]
print("Kolumny z powierzchnia:", area_cols)
```

```python
# wykres punktowy: powierzchnia mieszkalna vs cena
plt.figure(figsize=(7, 4))
plt.scatter(df['GrLivArea'], df['SalePrice'], alpha=0.5)
plt.xlabel('GrLivArea (powierzchnia mieszkalna nad ziemią)')
plt.ylabel('SalePrice')
plt.title('GrLivArea vs SalePrice')
plt.show()
```

Zależność jest wyraźna i liniowa — im większa powierzchnia mieszkalna, tym wyższa cena domu.  
Widać dwa dziwne punkty na dole po prawej (GrLivArea > 4500 sq ft, a cena poniżej 200k $) — to ewidentne outliery (nietypowe transakcje lub domy w złym stanie).

```python
# czy pow. dzialki (LotArea) ma taki sam wplyw jak pow. domu?
plt.figure(figsize=(7, 4))
plt.scatter(df['LotArea'], df['SalePrice'], alpha=0.5, color='green')
plt.xlabel('LotArea (powierzchnia działki)')
plt.ylabel('SalePrice')
plt.title('LotArea vs SalePrice')
plt.show()
```

Powierzchnia działki (`LotArea`) ma znacznie słabszy wpływ niż metraż domu. Jest sporo wielkich działek, które wcale nie kosztują dużo. Kupujący płacą głównie za sam budynek.

```python
# porownanie 3 rodzajow powierzchni
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].scatter(df['GrLivArea'], df['SalePrice'], alpha=0.4)
axes[0].set_title(f"GrLivArea (r = {df['GrLivArea'].corr(df['SalePrice']):.2f})")
axes[0].set_xlabel('GrLivArea')
axes[0].set_ylabel('SalePrice')

axes[1].scatter(df['TotalBsmtSF'], df['SalePrice'], alpha=0.4, color='orange')
axes[1].set_title(f"TotalBsmtSF (r = {df['TotalBsmtSF'].corr(df['SalePrice']):.2f})")
axes[1].set_xlabel('TotalBsmtSF')

axes[2].scatter(df['GarageArea'], df['SalePrice'], alpha=0.4, color='purple')
axes[2].set_title(f"GarageArea (r = {df['GarageArea'].corr(df['SalePrice']):.2f})")
axes[2].set_xlabel('GarageArea')

plt.tight_layout()
plt.show()
```

Najsilniejszy związek z ceną ma `GrLivArea` (r = 0.71), zaraz potem garaż `GarageArea` (r = 0.62) i piwnica `TotalBsmtSF` (r = 0.61).

```python
# dodatkowo: wplyw jakosci wykonczenia (OverallQual) jak na wykladzie
plt.figure(figsize=(8, 4))
sns.boxplot(x='OverallQual', y='SalePrice', data=df)
plt.title('OverallQual vs SalePrice')
plt.show()
```

## Wnioski
- Hipoteza się potwierdza: większe domy są droższe, przy czym decydujący jest metraż budynku (`GrLivArea`), a wielkość samej działki (`LotArea`) ma małe znaczenie.
- Rozkład cen jest skośny w prawo, średnia to ~180k, a mediana 163k.
- W danych są outliery (np. 2 ogromne domy sprzedane bardzo tanio i luksusowe rezydencje powyżej 340k).
- Poza metrażem kluczowa dla ceny jest jakość wykończenia (`OverallQual`).
